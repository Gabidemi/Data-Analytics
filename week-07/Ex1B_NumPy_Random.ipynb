```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1,20, size=2)
print(r_1)
```

```python
r_2 = np.random.randint(1,20, size=(1,2))
print(r_2)
```

```python
r_3 = np.random.rand(2,1)
print(r_3)
```

```python
r_4 = np.random.randint(20,100, size=(4,5))
print(r_4)
```

```python
r_5 = np.random.rand(4,5)
print(r_5)
```

## Generating random Choices from an Array

```python
students = np.array(['Ardita','Arber','Christopher','Denae','Djeneba','Idasha','Aalexus','Alvie'
                     ,'Richard','Ricardo','Arionsae','Deodat'])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3,4))
print(breakout_groups)

# This looks like a great randomized way to make breakout rooms because its random
```

```python
call_on_weighted = np.random.choice(students, p=[0.1, 0.1, 0.1, 0.3, 0.1, 0.1, 0.0, 0.1, 0.1, 0.0,
                                                 0.0, 0.0], size=5)
print(call_on_weighted)

# Not much to explain, I just randomly assigned them numbers but everything had to equal 1
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4,5))
print(bell_curve)
```
