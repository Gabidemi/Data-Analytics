# Exercise 3.A Getting Familiar with Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
```

## Creating Basic Plots

```python
x = [1,2,3,4,5]
y = [2,3,5,7,11]
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,marker='o')
plt.show()
```

```python
plt.plot(x,y,marker='o')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y label")
plt.show()
```

## Styling Plots

```python
plt.plot(x,y,marker='o',ls='--')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y label")
plt.show()
```

```python
plt.plot(x,y,marker='o',ls='--', c='r')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y label")
plt.show()
```

```python
plt.plot(x,y,marker='s', ms=10, mfc='g',ls='--', c='r')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y label")
plt.show()
```

```python
plt.plot(x,y,marker='*', ms=10, mfc='r',ls='--', c='b')
plt.title("Example Graph using star marker")
plt.xlabel("X axis")
plt.ylabel("Y label")
plt.show()
```

```python
plt.plot(x,y,marker='X', ms=10, mfc='r',ls='--', c='b')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y label")
plt.show()
```

# Exercise 3.B More Plotting practice

## Advanced Plotting Techniques

```python
products = ['Product A','Product B','Product C','Product D']
sales = [23,17,35,29]

plt.bar(products,sales)
plt.show()
```

```python
plt.bar(products,sales, color='skyblue', width=0.5)
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x,y)
plt.show()
```

```python
plt.scatter(x,y, s=np.random.rand(50)*100, cmap='viridis')
plt.show()
```

## Subplots

```python
plt.subplot(1,2,1)
x = np.linspace(0,10,100)
y = np.sin(x)

plt.plot(x,y)
plt.title("Sine Wave")
plt.xlabel("X value")
plt.ylabel("Y value")

plt.subplot(1,2,2)
random_numbers = np.random.randn(1000)

plt.hist(random_numbers)
plt.title("Histogram")
plt.xlabel("X value")
plt.ylabel("Y value")
plt.show()
```
