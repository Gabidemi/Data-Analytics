## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
```

```python
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
```

```python
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1,joined_array_1), axis=0)
```

```python
print(stacked_array_1)
```

```python
# another_2 = np.concatenate((joined_array_1,arrA), axis=1)

# A ValueError is displayed telling us that the input array dimensions must match
```

```python
testing_stack_1 = np.hstack((arrA,arrA))
print(testing_stack_1)

# THis displays each row and column in a horizontal line
```

```python
testing_stack_2 = np.vstack((arrA,arrA))
print(testing_stack_2)

# This one displays each row and column stacked on top of the other one
```

```python
testing_stack_3 = np.dstack((arrA,arrA))
print(testing_stack_3)

# It looks like it stacks the rows and column in a straight line downward and then
# the 2nd dimension is on the right of the first dimension
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)

# There are 2 arrays
# The dimensions are 2 dimensions each
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)

# Its split into 3 arrays 
# 2 different dimensions, the first one is 2D and the 2nd and 3rd ones are 1D
```

## Searching and Replacing Array values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'Odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([
    [7,6],
    [13,5],
    [7,3]
])
print(arrB)
```

```python
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = np.where(arrB < 7)
```

```python
arrC = arrB[z]
print(arrC)
```

```python
y = np.where(arrB > 7)
arrD = arrB[y]
print(arrD)
```
