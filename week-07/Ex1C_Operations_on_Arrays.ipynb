## Rounding

```python
print(r_3)
truncated = np.trunc(r_3)
fixed = np.fix(r_3)
floored = np.floor(r_3)
ceilinged = np.ceil(r_3)
rounded = np.round(r_3, 2)

#Using Trunc
print("Truncated")
print(truncated)

#Using fix
print("Fixed")
print(fixed)

#using floor
print("Floored")
print(floored)

#Using Ceiling
print("Ceilinged")
print(ceilinged)

#Rounded
print("Rounded")
print(rounded)
```

## Cumulative Operations

```python
print(r_4)
add_all = np.sum(r_4)
add_axis0 = np.sum(r_4, axis=0)
add_axis1 = np.sum(r_4, axis=1)
add_cumsum = np.cumsum(r_4)
add_cumsum0 = np.cumsum(r_4, axis=0)

#Added all the sums
print("Added Sums of all")
print(add_all)

#Adds the values vertically
print("Added Sums of axis 0")
print(add_axis0)

#Adds the values horizontally
print("Added Sums of axis 1")
print(add_axis1)


print("Added cumulative Sums of all")
print(add_cumsum)


print("Added cumulative Sums of all along 0 axis")
print(add_cumsum0)
```

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])

print(bag1)
print(bag2)
```

```python
print(f"The first bag contains {np.sum(bag1)} items," 
f"and the second bag contains {np.sum(bag2)} items.")
```

```python
total = bag1 + bag2

print(f"There are {total[0]} vegetables, {total[1]} fruits, {total[2]} boxes of snacks," 
f"and {total[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
Ali = np.array([5,4,6,4,2])
Ben = np.array([0,5,5,3,4])
Carla = np.array([3,6,0,3,4])

all_knocks = np.array([Ali, Ben, Carla])
print(all_knocks)
```

```python
AliWeek = np.sum(all_knocks, axis=1)[0]
print(f"Ali knocked on {AliWeek} doors this week")
```

```python
BenWeek = np.sum(all_knocks, axis=1)[1]
print(f"Ben knocked on {BenWeek} doors this week")
```

```python
CarlaWeek = np.sum(all_knocks, axis=1)[2]
print(f"Carla knocked on {CarlaWeek} doors this week")
```

```python
TotalWeek = np.sum([Ali, Ben, Carla])
print(f"In total, they knocked on {TotalWeek} doors this week")
```

```python
daily_knocks = np.sum(all_knocks, axis=0)
rolling_knocks = np.cumsum(daily_knocks)
estimated_votes = rolling_knocks * 0.5

print(estimated_votes)
```
