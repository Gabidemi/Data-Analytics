This is Lab 1 of Exercise 2.C. It demonstrates how to define functions, pass arguments, and return results.

```python
def favorite_things(name, movie, music):
    return "Name: " + name + "\nFavorite movie: " + movie + "\nFavorite musician/band: " + music

print(favorite_things("Gbolahan", "Bubble", "BabyMetal"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined Year Up United to develop my data analysis skills "
"and prepare for a career in the tech and financial industry."))
```

```python
def favorite_place(place, why):
    return "One of my favorite places is " + place + " because " + why + "."
```

```python
print(favorite_place("The Arcade", "its one of the best places to hang out and have fun"))
```

```python
print(favorite_place("My house", "because I can be lazier inside"))
```
