# Learning about Files
Reading and Writing from Files (The Basics)

### Creating a file and writing to a file


If the file does not exist, it creates a new file, then with file.write(), you can write into that file

```python
with open("Students.txt","w") as file:
    file.write('Alice\n')
    file.write('Bob\n')
```

### Reading from a file

If the file exists, you can read the information from that file and print it to a console

```python
with open("Students.txt", "r") as file:
    content = file.read()
    print(content)
```
