# Exercise 2.A Using built-in functions

## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):
    for i in range(len(part_code)):
        if part_code[i] == ":":
            return part_code[:i]
```

```python
def get_product_num(part_code):
    product_start = 0

    for i in range(len(part_code)):
        if part_code[i] == ":":
            product_start = i + 1

        if part_code[i] == "-":
            return part_code[product_start:i]
```

```python
def get_size(part_code):
    for i in range(len(part_code)):
        if part_code[i] == "-":
            return part_code[i + 1:]
```

```python
part_code1 = "ACME:123-L"
part_code2 = "DI:12345-M"
part_code3 = "ACE:1-12"
```

```python
print(
    "For " + part_code1
    + ", the supplier is " + get_supplier_code(part_code1)
    + ". Part number " + get_product_num(part_code1)
    + " is available in size " + get_size(part_code1) + "."
)
```

```python
print(
    "For " + part_code2
    + ", the supplier is " + get_supplier_code(part_code2)
    + ". Part number " + get_product_num(part_code2)
    + " is available in size " + get_size(part_code2) + "."
)
```

```python
print(
    "For " + part_code3
    + ", the supplier is " + get_supplier_code(part_code3)
    + ". Part number " + get_product_num(part_code3)
    + " is available in size " + get_size(part_code3) + "."
)
```
